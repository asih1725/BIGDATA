```python
# jawaban 2
# 1. Menu Kopi dengan tingkat kepahitan di atas 3
print("Menu Kopi dengan tingkat kepahitan di atas 3:")

for menu in koleksi_menu.find({
    "kategori": "Kopi",
    "tingkat_kepahitan": {"$gt": 3}
}, {"_id": 0}):
    print(menu)


# 2. Menu dengan harga antara 15000 dan 30000
print("\nMenu dengan harga Rp15.000 - Rp30.000:")

for menu in koleksi_menu.find({
    "harga": {
        "$gte": 15000,
        "$lte": 30000
    }
}, {"_id": 0}):
    print(menu)


# 3. Menu yang tidak memiliki field pedas_level
print("\nMenu yang tidak memiliki field pedas_level:")

for menu in koleksi_menu.find({
    "pedas_level": {"$exists": False}
}, {"_id": 0}):
    print(menu)
```

```python
# jawaban 3
# Menambahkan status Tersedia ke seluruh menu
hasil_update = koleksi_menu.update_many(
    {},
    {"$set": {"status": "Tersedia"}}
)

print("Jumlah dokumen yang diperbarui:",
      hasil_update.modified_count)


# Mengubah status Espresso menjadi Habis
koleksi_menu.update_one(
    {"nama_menu": "Espresso"},
    {"$set": {"status": "Habis"}}
)

# Mengubah status Nasi Goreng menjadi Habis
koleksi_menu.update_one(
    {"nama_menu": "Nasi Goreng"},
    {"$set": {"status": "Habis"}}
)


# Menampilkan seluruh menu setelah pembaruan
print("\nDaftar menu setelah update:")

for menu in koleksi_menu.find(
    {},
    {"_id": 0, "nama_menu": 1, "status": 1}
):
    print(menu)
```

```python
#jawaban 4
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg
import pandas as pd

# Membuat SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

# Langkah 1: Mengambil seluruh data dari MongoDB
dokumen_mongo = list(koleksi_menu.find({}))

# Langkah 2: Mengubah data menjadi Pandas DataFrame
df_pandas = pd.DataFrame(dokumen_mongo)

# Langkah 3: Menyesuaikan tipe data MongoDB
df_pandas["_id"] = df_pandas["_id"].astype(str)

# Mengubah nilai kosong (NaN) menjadi None
df_pandas = df_pandas.astype(object).where(
    pd.notna(df_pandas), None
)

# Langkah 4: Mengonversi menjadi Spark DataFrame
df_spark_menu = spark.createDataFrame(df_pandas)

# Menampilkan struktur DataFrame
print("Struktur Spark DataFrame:")
df_spark_menu.printSchema()

# Menampilkan seluruh data
print("Data Menu:")
df_spark_menu.show(truncate=False)

# Menghitung rata-rata harga berdasarkan kategori
print("Rata-rata Harga Menu per Kategori:")

df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
).show()

# Menutup SparkSession
spark.stop()

print("SparkSession ditutup.")
```
