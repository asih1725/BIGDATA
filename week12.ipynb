```python
import pandas as pd
import zipfile

zip_path = "/home/asih2/Downloads/archive(2).zip"

# 1️⃣ Buka ZIP dan baca tested.csv
with zipfile.ZipFile(zip_path, 'r') as z:
    with z.open("tested.csv") as f:
        df = pd.read_csv(f)

print("=== DATA AWAL ===")
print(df.head(), "\n")

# 2️⃣ Identifikasi missing values
print("=== CEK MISSING VALUES ===")
print(df.isnull().sum(), "\n")

# 3️⃣ Tentukan kolom numerik & kategorikal otomatis
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("Kolom numerik:", numeric_cols)
print("Kolom kategorikal:", categorical_cols, "\n")

# 4️⃣ Tangani missing values
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# 5️⃣ Normalisasi kolom numerik
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

# 6️⃣ Standarisasi kolom kategorikal
for col in categorical_cols:
    df[col] = df[col].astype(str).str.capitalize()

# 7️⃣ Hapus duplikat
df.drop_duplicates(inplace=True)

print("=== DATA SETELAH BERSIH ===")
print(df.head(), "\n")

# 8️⃣ Simpan hasil
output = "/home/asih2/Downloads/tested_cleaned.csv"
df.to_csv(output, index=False)

print("File berhasil disimpan ke:", output)
```

```python
# === IMPORT LIBRARY ===
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# === 1. LOAD DATASET DARI ZIP ===
zip_path = "/home/asih2/Downloads/archive(2).zip"   # Lokasi zip kamu
df = pd.read_csv(zip_path, compression="zip")
print("=== DATA AWAL ===")
print(df.head(), "\n")

# === 2. CEK MISSING VALUES ===
print("=== JUMLAH MISSING VALUES ===")
print(df.isnull().sum(), "\n")

# === 3. TANGANI MISSING VALUES ===
# Pisahkan kolom numerik dan kategorikal
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Isi missing numeric dengan mean
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Isi missing kategori dengan modus
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values setelah diperbaiki:")
print(df.isnull().sum(), "\n")

# === 4. HANDLE OUTLIERS (IQR METHOD) ===
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    # Hapus baris outlier
    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("=== DATASET SETELAH OUTLIER DIBERSIHKAN ===")
print(df.describe(), "\n")

# === 5. NORMALISASI KOLUMN NUMERIK ===
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("=== DATA SETELAH NORMALISASI ===")
print(df.head(), "\n")

# === 6. ENCODING KATEGORIKAL (LABEL ENCODER) ===
encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col])

print("=== DATA SETELAH ENCODING ===")
print(df.head(), "\n")

# === 7. HAPUS DUPLIKAT ===
df.drop_duplicates(inplace=True)

print("Jumlah data akhir:", len(df), "\n")

# === 8. SAVE DATA BERSIH ===
df.to_csv("cleaned_tested.csv", index=False)
print("Dataset bersih berhasil disimpan sebagai cleaned_tested.csv")
```
